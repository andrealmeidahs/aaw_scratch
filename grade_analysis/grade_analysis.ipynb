```python
import os
import re
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
def find_empty_line_no(f):
    lno = 0
    with open(f, "r") as fid:
        for line in fid:
            if len(line) < 2:
                return lno
            lno += 1

def read_grade_file(f):
    data = {}
    with open(f, "r") as fid:
        for line in fid:
            if line.startswith("#"):
                continue
            if len(line) < 2:
                continue
            name, values = line.split(maxsplit=1)
            values_split = values.split()
            try:
                float(values_split[0])
                row = np.fromstring(values, sep=" ")
            except ValueError:
                row = values_split
            data[name] = row
    return pd.DataFrame(data)
```

```python
grade_dir = r"C:\Dev\Git\Woodsmith\conf\grade\grade_v4\quarters_17_40_lenCV"
```

```python
i = 0
all_df = []
for root, dirs, files in os.walk(grade_dir):
    for file in files:
        if file.endswith(".dat"):
            file_path = os.path.join(root, file)
            seed_no = int(re.search(r"seed(\d+)", file_path).group(1))
            cm_no = int(re.search(r"GradeCM(\d+)", file_path).group(1))
            empty_line_no = 3#find_empty_line_no(file_path)
            #df = pd.read_csv(file_path, skiprows=empty_line_no, header=None, sep=r'\s+', index_col=0, engine='c').T
            print(f"Reading {file_path} ...")
            df = read_grade_file(file_path)
            if min(df.shape) == 1:
                print(f"Skipping {file_path} due to insufficient data.")
                continue
            df["seed_no"] = seed_no
            df["cm_no"] = cm_no
            df["file_no"] = i
            
            all_df.append(df) 
            # Perform analysis on the DataFrame 'df'
            i+=1
```

```python
df = pd.concat(all_df, ignore_index=True)
```
